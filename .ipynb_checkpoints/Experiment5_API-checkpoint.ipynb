#  Exp 5 Heart Disease API Model

## Step 1 — Check dataset

```python
import pandas as pd
import os

dataset_path = "data/heart_disease_uci_cleaned.csv"

df = pd.read_csv(dataset_path)

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nTarget distribution:")
print(df["num"].value_counts().sort_index())

df.head()
```

## Step 2 — Prepare X and y

```python
# Separate features and target

X = df.drop(columns=["num", "id", "dataset"])
y = df["num"]

print("Feature columns:")
print(X.columns.tolist())

print("\nNumber of features:", X.shape[1])
print("Target:", "num")
```

## Step 3 — Train/Test Split

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

## Step 4 — Train Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

api_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=10,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42
)

api_model.fit(X_train, y_train)

print("Heart Disease Random Forest model trained successfully!")
```

## Step 5 — Evaluate the API model

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

y_pred = api_model.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("===== Heart Disease API Model Evaluation =====")
print("Accuracy :", round(accuracy, 6))
print("Precision:", round(precision, 6))
print("Recall   :", round(recall, 6))
print("F1 Score :", round(f1, 6))

print("\nClassification Report:")
print(classification_report(y_test, y_pred, zero_division=0))
```

## Step 6 — Save the model

```python
import joblib
import os

os.makedirs("models", exist_ok=True)

model_path = "models/heart_disease_api_model.joblib"

joblib.dump(api_model, model_path)

print("Model saved successfully!")
print("Path:", model_path)
```

## Step 7 — Verify saved model

```python
import os

print("Model exists:", os.path.exists(model_path))
print("Model size:", os.path.getsize(model_path), "bytes")
```

## Step 8 — Save model metadata

```python
import json

model_info = {
    "model_name": "HeartDisease_4373_API_Model",
    "model_version": "1.0",
    "algorithm": "Random Forest",
    "n_estimators": 100,
    "max_depth": 10,
    "number_of_features": 13,
    "target": "num",
    "accuracy": round(accuracy, 6),
    "precision_weighted": round(precision, 6),
    "recall_weighted": round(recall, 6),
    "f1_weighted": round(f1, 6)
}

with open("models/model_info.json", "w") as f:
    json.dump(model_info, f, indent=4)

print("Model metadata saved successfully!")
```
